# Melody Extraction Pipeline

Run this notebook from top to bottom to prepare an audio file, split it into sections, run the configured pitch models, and export pitch tracks.


## 1. Inspect PitchLab Models


```python
import pitchlab
import os
#print(pitchlab.available_models())
```

## 2. Select Input Audio


```python
file_name = "uzbek_dari.mp3"
file_path = "../experiments/audio_samples/"+file_name
```

## 3. Filter Audio And Preview


```python
import librosa
import soundfile as sf
import numpy as np
from scipy.signal import butter, sosfiltfilt
from IPython.display import Audio
from tools.audio_modifiers import high_pass, segment_audio

print(high_pass.__code__.co_varnames)

temp_path = "temp" + file_name[-4:]
ENABLE_FILTER = True

# Load audio as mono, preserving original sample rate
audio, sr = librosa.load(file_path, sr=None, mono=True)

filtered_audio = high_pass(
    audio=audio,
    sr=sr,
    cutoff_hz=147,
    order=10,
    output_path=temp_path
)

Audio(filename=temp_path)
```

## 4. Split The Filtered Audio Into Sections


```python
from IPython.display import Audio, display, Markdown
#splice files into different sections

splices = [0,27, 43, 56, 78, 89, -1]
segment_audio(temp_path, splices)

print(len(splices))
for i in range(len(splices)):
    temp_path = f"section_{i}.mp3"
    print(temp_path)
    #display(Audio(temp_path, rate=44100))
```

## 5. Run Pitch Models On One Section


```python
temp_path = f"section_{1}.mp3"
from pitchlab.audio_io import load_audio
from tqdm import tqdm
audio, sr = load_audio(
    temp_path
)

results_df = []
model_list = ["fcpe","essentia-melodia","pyworld-dio","librosa-yin","yaapt","librosa-pyin", "torchcrepe"]
for model_name in tqdm(model_list):
    if model_name != "crepe": #redneck fix 
        try:
            print(f"attempting to run {model_name}")
            model = pitchlab.get_model(model_name)
            df = model.predict(
                audio,
                sr,
                range=(200, 1007),
            )
            results_df.append(df)
        except Exception as e:
            print("Model Failed To Run")
            print(e)

#os.remove(temp_path)
```

## 6. Export Pitch Runs


```python
from tools.pitch_run_io import export_pitch_runs

output_dir = "../experiments/pitch_runs/"
saved_paths = export_pitch_runs(results_df, output_dir, file_name)

for path in saved_paths:
    print(f"saved {path}")
```
