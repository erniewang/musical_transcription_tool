# Individual Model Smoke Tests

Run the setup cell first. Then run each model cell individually to see whether that model imports and handles a tiny sine-wave audio input.

```python
from pathlib import Path
import sys
import traceback

import numpy as np
from IPython.display import Audio, display

cwd = Path.cwd().resolve()
source_root = None
for base in (cwd, *cwd.parents):
    candidates = [
        base,
        base / "melody_extractor" / "pitchlab-ernie",
    ]
    for candidate in candidates:
        if (candidate / "pitchlab" / "model_folders").is_dir():
            source_root = candidate
            break
    if source_root is not None:
        break

if source_root is None:
    raise RuntimeError("Could not find the pitchlab source package from the current working directory.")

sys.path.insert(0, str(source_root))

SAMPLE_RATE = 22050
DURATION_SECONDS = 1.0
FREQUENCY_HZ = 440.0
time = np.linspace(0.0, DURATION_SECONDS, int(SAMPLE_RATE * DURATION_SECONDS), endpoint=False)
audio = (0.2 * np.sin(2.0 * np.pi * FREQUENCY_HZ * time)).astype(np.float32)
params = {
    "sample_rate": SAMPLE_RATE,
    "fmin": 50.0,
    "fmax": 1000.0,
    "hop_ms": 10.0,
}

def smoke_test(model_name, runner, extra_params=None):
    run_params = dict(params)
    if extra_params:
        run_params.update(extra_params)
    print(f"Running {model_name}...")
    try:
        result = runner(audio, run_params)
        print(f"{model_name} OK")
        print(f"rows: {len(result)}")
        display(result.head())
        return result
    except Exception:
        print(f"{model_name} FAILED")
        traceback.print_exc()

display(Audio(audio, rate=SAMPLE_RATE))
print(f"Using pitchlab package from: {source_root}")
print(f"Generated {DURATION_SECONDS:g}s test tone at {FREQUENCY_HZ:g} Hz, sample rate {SAMPLE_RATE} Hz.")
```
