```python
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.io as pio

pio.renderers.default = "notebook"

base = Path("../experiments/pitch_runs")

#show every time series graph
for file in base.rglob("*.csv"):
    df = pd.read_csv(file).fillna(0)

    fig = px.line(
        df,
        x="time_sec",
        y="f0_hz",
        title=file.stem
    )
    #fig.show()
```

```python
from tqdm import tqdm
base = Path("../experiments/pitch_runs")

data_frames = []
for folder in base.iterdir():
    if folder.is_dir():
        for file in folder.iterdir():
            if file.suffix == ".csv":
                df = pd.read_csv(file).fillna(0)
                data_frames.append(df)

for df in data_frames:
    print(df[:10])
# model  time_sec  f0_hz  confidence  voiced
```

```python
from functools import reduce

cleaned = []

for i, df in enumerate(data_frames):
    df = df.drop(columns=["voiced", "model"], errors="ignore").copy()

    df = df.rename(columns={
        col: f"{col}_{i}"
        for col in df.columns
        if col != "time_sec"
    })
    
    cleaned.append(df)

aggregate = reduce(
    lambda left, right: pd.merge(left, right, on="time_sec", how="outer"),
    cleaned
).fillna(0)

#this is retarded, you cant access a single index you get key error but if you do [x:] or [:y] then it works
#pandas dataframe is actually not a list. and the slicing stuff is just put in there. 
print(aggregate.iloc[70])
```
